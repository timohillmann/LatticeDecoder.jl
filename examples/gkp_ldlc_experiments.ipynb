```sage

def symplectic_form(n):
    """
    Creates the standard symplectic form for n qubits.
    
    Args:
    n (int): Number of qubits
    
    Returns:
    numpy.ndarray: 2n x 2n symplectic form matrix
    """
    zero_block = np.zeros((n, n), dtype=int)
    identity_block = np.eye(n, dtype=int)
    
    top_row = np.hstack((zero_block, identity_block))
    bottom_row = -1*np.hstack((identity_block, zero_block))
    
    return np.vstack((top_row, bottom_row))
```
